## 1. Student Information


```python
student_name = "Usman"
age = 25
city = "Karachi"
is_student = True

print("Student Name:", student_name)
print("Age:", age)
print("City:", city)
print("Student Status:", is_student)
```

## 2. Data Types


```python
text_value = "Python"
integer_value = 25
float_value = 5.6
boolean_value = True

print(text_value, type(text_value))
print(integer_value, type(integer_value))
print(float_value, type(float_value))
print(boolean_value, type(boolean_value))
```

## 3. Arithmetic Operations


```python
num1 = 20
num2 = 5

print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
print("Modulus:", num1 % num2)
print("Power:", num1 ** num2)
```

## 4. Student Marks Calculator


```python
math_marks = 80
english_marks = 70
computer_marks = 90

total_marks = math_marks + english_marks + computer_marks
average_marks = total_marks / 3

print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
```

## 5. Shopping Bill


```python
product_price = 500
quantity = 3

total_bill = product_price * quantity

print("Product Price:", product_price)
print("Quantity:", quantity)
print("Total Bill:", total_bill)
```

## 6. Rectangle Area


```python
length = 10
width = 5

area = length * width
perimeter = 2 * (length + width)

print("Area:", area)
print("Perimeter:", perimeter)
```

## 7. Time Converter


```python
minutes = 250

hours = minutes // 60
remaining_minutes = minutes % 60

print("Hours:", hours)
print("Remaining Minutes:", remaining_minutes)
```

## 8. Student Eligibility


```python
marks = 65
attendance = 80

eligible = marks >= 50 and attendance >= 75

print("Marks:", marks)
print("Attendance:", attendance)
print("Eligible:", eligible)
```

## 9. Logical Operators


```python
a = True
b = False

print("a and b:", a and b)
print("a or b:", a or b)
print("not a:", not a)
print("not b:", not b)
```
